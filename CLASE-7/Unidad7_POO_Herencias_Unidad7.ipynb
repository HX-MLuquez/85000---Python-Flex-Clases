## 1) Actividad: Diagrama de Clase -> Implementación en Python

**Consigna.** Crear una clase `Alumno` con atributos de instancia `nombre` y `nota` (públicos). Métodos: `__init__`, `imprimir()` y `resultado()`.

**Aclaración:** `imprimir()` mostrará por pantalla el nombre y la nota. `resultado()` evalúa la nota: si < 5 `desaprueba`, si > 5 `aprueba`. Añadimos un caso para `== 5` para explicar el empate.

A continuación: implementación y ejemplos con al menos dos instancias.

```
class Alumno:
    """Clase Alumno simple con métodos públicos:
    - nombre: str
    - nota: float/int
    - imprimir(): imprime nombre y nota
    - resultado(): devuelve si aprueba o desaprueba (y caso límite)
    """
    def __init__(self, nombre, nota):
        self.nombre = nombre
        self.nota = nota

    def imprimir(self):
        print(f"Nombre: {self.nombre} — Nota: {self.nota}")

    def resultado(self):
        # Interpretación explícita de la consigna:
        # si nota < 5 => desaprueba
        # si nota > 5 => aprueba
        # si nota == 5 => caso límite (puede requerir recuperación o revisión)
        if self.nota < 5:
            return "Desaprueba"
        elif self.nota > 5:
            return "Aprueba"
        else:
            return "Resultado límite: exactamente 5 (consultar reglamento)"
```

```
# Instanciamos al menos dos objetos y probamos los métodos
alumno1 = Alumno('Lucía', 7.5)
alumno2 = Alumno('Mateo', 4.0)
alumno3 = Alumno('Carla', 5.0)  # para ver el caso límite

for a in (alumno1, alumno2, alumno3):
    a.imprimir()
    print('->', a.resultado())
    print('---')
```

## 2) Repaso: Relación entre clases (ejemplo realista — Club)

Explicación y ejemplo usando `Socio`, `Profesor`, `Deporte`, `Actividad`, `Club`. Veremos distintos tipos de relaciones:

- **Asociación**: un `Profesor` puede dictar varias `Actividades`, una `Actividad` tiene un `Profesor`.
- **Composición**: un `Club` contiene `Actividades` y `Socios` (si se borra el `Club`, las actividades asociadas dejan de tener sentido en ese contexto).
- **Agregación**: un `Socio` puede pertenecer a varios `Clubes` (vida propia fuera del club).

A continuación código que genera estas relaciones y muestra varios `print` que ayudan a entender qué se representa.

```
class Profesor:
    def __init__(self, nombre, especialidad):
        self.nombre = nombre
        self.especialidad = especialidad

    def __repr__(self):
        return f"Profesor({self.nombre}, {self.especialidad})"

class Deporte:
    def __init__(self, nombre):
        self.nombre = nombre

    def __repr__(self):
        return f"Deporte({self.nombre})"

class Actividad:
    def __init__(self, nombre, deporte, profesor):
        self.nombre = nombre
        self.deporte = deporte
        self.profesor = profesor

    def __repr__(self):
        return f"Actividad({self.nombre}, {self.deporte.nombre}, {self.profesor.nombre})"

class Socio:
    def __init__(self, nombre, socio_id):
        self.nombre = nombre
        self.socio_id = socio_id

    def __repr__(self):
        return f"Socio({self.nombre}, id={self.socio_id})"

class Club:
    def __init__(self, nombre):
        self.nombre = nombre
        self.socios = []         # composición/colección de socios
        self.actividades = []    # composición/colección de actividades

    def agregar_socio(self, socio):
        self.socios.append(socio)

    def agregar_actividad(self, actividad):
        self.actividades.append(actividad)

    def __repr__(self):
        return f"Club({self.nombre})"

# Creamos algunos objetos reales
prof1 = Profesor('Ana', 'Natación')
prof2 = Profesor('Jorge', 'Fútbol')

natacion = Deporte('Natación')
futbol = Deporte('Fútbol')

act1 = Actividad('Mañana natación', natacion, prof1)
act2 = Actividad('Tarde fútbol', futbol, prof2)

s1 = Socio('María', 101)
s2 = Socio('Pablo', 102)

club = Club('Club Las Palmeras')
club.agregar_socio(s1)
club.agregar_socio(s2)
club.agregar_actividad(act1)
club.agregar_actividad(act2)

# Prints ilustrativos (los "últimos print" que habría que explicar en clase)
print('Club:', club.nombre)
print('Socios del club:', club.socios)
print('Actividades del club:', club.actividades)
print('Profesor de la primera actividad:', club.actividades[0].profesor)
print('¿El mismo profesor puede dar más de una actividad? ->', prof1, 'puede dar múltiples actividades si se lo asignamos.')
```

### ¿Cómo explicar los últimos `print`?

- `Club: Club Las Palmeras` muestra el objeto `Club` en su forma más simple.
- `Socios del club:` imprime la lista completa de instancias `Socio`. Cada socio es un objeto independiente.
- `Actividades del club:` imprime la lista de `Actividad` con referencia al `Deporte` y al `Profesor` asociados.
- `Profesor de la primera actividad:` muestra la referencia al objeto `Profesor` que está dentro de `Actividad` —esto es **asociación**.

**Analogía**: Piensa en el `Club` como un edificio: dentro hay salas (actividades) y personas (socios). Los `Profesores` son trabajadores que pueden estar asignados a salas; si un profesor trabaja en dos salas, lo vemos en dos actividades (asociación compartida).

## 3) Repaso: Para pensar… (Pista: atención a los parámetros por defecto y atributos compartidos)

Aquí mostramos un clásico error: usar una lista como valor por defecto en el constructor produce que **todos** los objetos compartan la misma lista, salvo que la creemos en `__init__` correctamente.

```
# Ejemplo problemático (no hacer así)
class ClubMalo:
    def __init__(self, nombre, socios=[]):  # ⚠️ socios=[] se comparte entre instancias
        self.nombre = nombre
        self.socios = socios

c1 = ClubMalo('A')
c2 = ClubMalo('B')
c1.socios.append('SocioX')
print('Club A socios:', c1.socios)
print('Club B socios (INESPERADO):', c2.socios)  # muestra el mismo cambio: sorpresa!

# Forma correcta
class ClubBueno:
    def __init__(self, nombre, socios=None):
        self.nombre = nombre
        self.socios = [] if socios is None else socios

cb1 = ClubBueno('A')
cb2 = ClubBueno('B')
cb1.socios.append('SocioX')
print('\nForma correcta:')
print('Club A socios:', cb1.socios)
print('Club B socios:', cb2.socios)
```

**Explicación**: El parámetro por defecto `[]` se evalúa una vez en definición de la función. Por eso, si se muta, todos los objetos que usan ese valor por defecto reciben la mutación. Es una trampa clásica en Python.

## 4) Ejemplo en vivo: Herencia (Animal) y Herencia múltiple

Definimos una `class Animal` con atributos y métodos genéricos. Luego especializamos en `Mamifero`, `AnimalMarino`, `Cetaceo` y creamos una clase que combine varios padres para mostrar la herencia múltiple y el MRO (Method Resolution Order).

Usaremos además una analogía: **Club** = clase base; **Secciones del Club** = subclases que añaden comportamientos.

```
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre

    def mover(self):
        return f"{self.nombre} se mueve (comportamiento genérico)"

class Mamifero(Animal):
    def __init__(self, nombre, tiene_pelo=True):
        super().__init__(nombre)
        self.tiene_pelo = tiene_pelo

    def alimentar(self):
        return f"{self.nombre} alimenta a sus crías con leche (característica mamífera)."

    def mover(self):
        return f"{self.nombre} camina o corre (mamífero)."  # sobrescribe mover

class AnimalMarino:
    def __init__(self, vive_en_agua=True):
        self.vive_en_agua = vive_en_agua

    def respirar_agua(self):
        return "Puede adaptarse al medio acuático (depende de la especie)."


class Cetaceo:
    def bucear(self):
        return "Puede bucear y nadar eficientemente."

    def mover(self):
        return "Se propulsa mediante aletas (comportamiento de cetáceo)."
```

```
# Herencia múltiple: Delfín hereda de Mamífero, Cetáceo y AnimalMarino
class Delfin(Mamifero, Cetaceo, AnimalMarino):
    def __init__(self, nombre):
        # Notar: como AnimalMarino y Cetaceo no comparten la misma jerarquía con super(),
        # llamamos explícitamente a los iniciadores relevantes.
        Mamifero.__init__(self, nombre, tiene_pelo=False)  # delfines no tienen pelo visible
        AnimalMarino.__init__(self, vive_en_agua=True)

    # no definimos mover() aquí para que se vea qué método toma el MRO
    def __repr__(self):
        return f"Delfin({self.nombre})"

d = Delfin('Bobby')
print(d)
print('alimentar:', d.alimentar())
print('bucear:', d.bucear())
# ¿qué mover se usa? -> según MRO (Mamifero, Cetaceo, AnimalMarino)
print('mover (MRO):', d.mover())
print('MRO de Delfin:', [c.__name__ for c in Delfin.__mro__])
```

### Explicación de la herencia múltiple y MRO

- `Delfin.__mro__` muestra el orden en que Python buscará métodos. En nuestro ejemplo, `mover()` es tomado de `Mamifero` porque `Mamifero` aparece primero en la lista de herencia (`class Delfin(Mamifero, Cetaceo, AnimalMarino)`).

**Analogía**: imagina que un `Delfín` es una persona que trabaja en dos departamentos del club: si pide instrucciones, primero pregunta al jefe del primer departamento (Mamífero), si no está, pasa al segundo (Cetáceo).

## 5) Actividad práctica: Herencia múltiple aplicada a contexto de Club

**Consigna.** Crear una herencia múltiple con `Mamífero`, `Cetáceo`, `AnimalMarino` —ya lo hicimos arriba con `Delfin`.

Aquí damos otro ejemplo más cercano al club/roles: imagine `Persona` -> `Profesor`, `Entrenador`, `Administrativo`. Una persona puede tener múltiples roles (herencia múltiple). A continuación un ejemplo sencillo y un consejo: en dominios reales, preferir composición sobre herencia múltiple para roles que cambian dinámicamente (por ejemplo, un `Socio` que también sea `Profesor` se modela mejor con roles asignables).

```
# Ejemplo de 'roles' con herencia múltiple (didáctico)
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

    def identidad(self):
        return f"Soy {self.nombre}"

class ProfesorRol(Persona):
    def dar_clase(self):
        return f"{self.nombre} da una clase."

class AdministrativoRol(Persona):
    def gestionar(self):
        return f"{self.nombre} gestiona recursos."

class ProfesorAdministrativo(ProfesorRol, AdministrativoRol):
    pass

pa = ProfesorAdministrativo('Sofía')
print(pa.identidad())
print(pa.dar_clase())
print(pa.gestionar())
print('MRO:', [c.__name__ for c in ProfesorAdministrativo.__mro__])
```

**Comentario final y buenas prácticas**:

- Para modelar roles (por ejemplo, un socio que también da clases), **composición** (tener una lista de roles o componentes) suele ser más flexible que mezclar muchas herencias.
- Use herencia cuando hay **is-a** claro (un `Delfín` es un `Mamífero`). Use composición cuando hay `has-a` o `roles` dinámicos (un `Socio` tiene un `RolProfesor` si corresponde).
