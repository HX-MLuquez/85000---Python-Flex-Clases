```python
class Alumno:

    def __init__(self,nombre,nota, id=None):
        self.nombre=nombre
        self.nota=nota
        self.id=id

    # función para imprimir los datos
    def imprimir(self):
               print("Nombre: ",self.nombre)
               print("Nota: ",self.nota)
 
 
    # función para obtener el resultado
    def resultado(self):
               if self.nota < 5:
                              print("El alumno ha suspendido")
               else:
                              print("El alumno ha aprobado")
                              
    #Método especial __str__
    def __str__(self):
        return self.id.__str__()
  
#bloque principal
 
alumno1 =Alumno("ivan",8, 1234)
alumno2 =Alumno("juan",4, 3333)
 
# imprimimos los datos y resultados de cada alumno
alumno1.imprimir()
alumno1.resultado()
alumno2.imprimir()
alumno2.resultado()

print(alumno1)
```
